# Import Required Libraries
This section imports the necessary libraries for working with SpaCy, visualization, and the test data conversion utilities.

```python
import spacy
from spacy.tokens import DocBin
from spacy import displacy
import os
from pathlib import Path

```

# Load Test Data
Load a .spacy file generated from your test data. This file contains serialized SpaCy Doc objects.

```python
# Path to the test .spacy file (adjust as needed)
docbin_path = Path("..") / "conversion" / "trsic2-dev.spacy"

# Load the DocBin
nlp = spacy.blank("sr")  
doc_bin = DocBin().from_disk(str(docbin_path))
docs = list(doc_bin.get_docs(nlp.vocab))
print(f"Loaded {len(docs)} SpaCy Doc objects from {docbin_path}")
```

# Visualize SpaCy Docs
Use SpaCy's displacy to visualize named entities and dependency parses for the loaded documents.

```python
# Visualize named entities in the first document (if any)
if docs and docs[0].ents:
    displacy.render(docs[0], style="ent", jupyter=True)
else:
    print("No named entities found in the first document.")
```

```python
#examine sentences in first doc
for sent in docs[0].sents:
    print(sent.text)
    print('--- ')
```

# Inspect Named Entities
Programmatically inspect the named entities, their labels, and any linked knowledge base IDs in the SpaCy Docs.

```python
# Collect entity details from the first document
if docs:
    ents_data = [
        {
            "Text": ent.text,
            "Label": ent.label_,
            "KB_ID": getattr(ent, "kb_id_", None)
        }
        for ent in docs[0].ents
    ]
    if ents_data:
        for ent in ents_data:
            print(ent)
            print('---')
    else:
        print("No entities found in the first document.")
else:
    print("No documents loaded.")
```

```python
#show the first document text, taggs,lemmas and sentences
if docs:
    doc = docs[0]
    print(f"Text: {doc.text}")
    print(f"Tokens: {[token.text for token in doc]}")
    print(f"Lemmas: {[token.lemma_ for token in doc]}")
    print(f"POS Tags: {[token.tag_ for token in doc]}")
    print(f"Sentences: {[sent.text for sent in doc.sents]}")
```

```python
# Preview the first 10 documents (or all if fewer) for quick inspection
if docs:
    preview_count = min(10, len(docs))
    print(f"Previewing {preview_count} document(s)...")
    for idx, doc in enumerate(docs[:preview_count], start=1):
        sentences = list(doc.sents)
        sentence_count = len(sentences)
        print(f"\n--- Document {idx}/{preview_count} ---")
        print(f"Length: {len(doc)} tokens | Sentences: {sentence_count} | Entities: {len(doc.ents)}")
        snippet = doc.text[:300].replace("\n", " ")
        if len(doc.text) > 300:
            snippet += "..."
        print(f"Snippet: {snippet}")
        if sentences:
            print("Sentences:")
            for sent_idx, sent_text in enumerate(sentences[:3], start=1):
                print(f"  {sent_idx}. {sent_text}")
        else:
            print("No sentences available (check if sentencizer is enabled).")
        lemma_tag_preview = [
            {
                "token": token.text,
                "lemma": token.lemma_,
                "tag": token.tag_
            }
            for token in doc[:5]
        ]
        print(f"Lemma/Tag preview: {lemma_tag_preview}")
        if doc.ents:
            ent_preview = [f"{ent.text} ({ent.label_})" for ent in doc.ents[:5]]
            print(f"Entity preview: {ent_preview}")
        else:
            print("No entities in this document.")
else:
    print("No documents loaded.")
```

```python
# Inspect whitespace handling for the first document
if docs:
    doc = docs[0]
    print(f"Document Text (first 200 chars):\n{doc.text[:200]!r}")
    print("\nToken Whitespace Analysis (first 20 tokens):")
    print(f"{'Token':<20} | {'Has Space After':<15} | {'Next Token'}")
    print("-" * 50)
    for i in range(min(20, len(doc)-1)):
        token = doc[i]
        next_token = doc[i+1]
        has_space = token.whitespace_ != ""
        print(f"{token.text:<20} | {str(has_space):<15} | {next_token.text}")
else:
    print("No documents loaded.")
```
